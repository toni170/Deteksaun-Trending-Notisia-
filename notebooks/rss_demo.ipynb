# Demo: Parsing RSS Feed Tatoli dengan `feedparser`

Notebook ini mendemonstrasikan cara mengambil & parsing RSS feed Tatoli (https://tatoli.tl/feed/), lalu menjalankan tahap awal preprocessing dari pipeline `news_topic_mining`.

```python
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join('..')))

import feedparser
import pandas as pd

from src.scraping.rss_scraper import fetch_feed, save_raw
from src.preprocessing.pipeline import preprocess_dataframe
```

## 1. Parsing feed langsung dengan `feedparser` (eksplorasi struktur data)

```python
FEED_URL = "https://tatoli.tl/feed/"
parsed = feedparser.parse(FEED_URL)

print("Judul feed:", parsed.feed.get('title'))
print("Jumlah entri:", len(parsed.entries))

if parsed.entries:
    first = parsed.entries[0]
    print("\nContoh entri pertama:")
    print("Judul   :", first.get('title'))
    print("Link    :", first.get('link'))
    print("Tanggal :", first.get('published'))
    print("Ringkasan:", first.get('summary', '')[:200])
```

## 2. Menggunakan modul `rss_scraper` (fetch + simpan ke `data/raw`)

```python
entries = fetch_feed(FEED_URL, source_name="Tatoli")
df_raw = pd.DataFrame(entries)
df_raw.head()
```

```python
paths = save_raw(entries, raw_dir="../data/raw")
paths
```

## 3. Cek hasil preprocessing (cleaning + deteksi bahasa + stemming)

```python
cleaned_df = preprocess_dataframe(df_raw)
cleaned_df[["title", "language", "text_final"]].head(10)
```
